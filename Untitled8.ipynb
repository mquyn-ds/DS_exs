```python
df['Tốc độ trải nghiệm'].value_counts().sort_index().plot(kind="pie", autopct='%1.1f%%')

plt.title("Tốc độ trải nghiệm")
plt.show()
```

```python
corr = df.corr(numeric_only=True)

print(corr["Mức độ hài lòng tổng thể"])
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

sns.heatmap(corr, annot=True)
plt.show()
```

```python
df["Mức độ hài lòng tổng thể"].value_counts().sort_index().plot(kind="bar", color = 'green')

plt.title("Mức độ hài lòng tổng thể")
plt.xlabel("Mức đánh giá")
plt.ylabel("Số lượng sinh viên")
plt.show()
```

```python
from sklearn.feature_extraction.text import CountVectorizer #bonus

texts = df["Sự cố kỹ thuật"].dropna()

vectorizer = CountVectorizer(
    ngram_range=(2,3), # cụm 2-3 từ
    stop_words=None
)

X = vectorizer.fit_transform(texts)

sum_words = X.sum(axis=0)

words_freq = [
    (word, sum_words[0, idx])
    for word, idx in vectorizer.vocabulary_.items()
]

print(words_freq)
```
